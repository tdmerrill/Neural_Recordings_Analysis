```python
def load_neurons_from_database(db_path, select_columns, conditions, table):
    """
       load data from sql database that fits given conditions

       Parameters
       ----------
       select_columns : list
           List of neuron features to return
       conditions : {'feature', 'condition'}
           conditions for neurons that will be returned (ex. manual_isi_1 < 1)

       Returns
       -------
       neurons : np.ndarray
           neurons that fit the criteria
       """

    import sqlite3

    conn = sqlite3.connect(db_path)
    cursor = conn.cursor()

    select_sql = ", ".join(select_columns)

    query = f"SELECT {select_sql} FROM {table}"
    values = []

    if conditions:
        where_clauses = []

        for col, (op, val) in conditions.items():
            where_clauses.append(f"{col} {op} ?")
            values.append(val)

        where_sql = " AND ".join(where_clauses)
        query += f" WHERE {where_sql}"

    cursor.execute(query, values)

    rows = cursor.fetchall()

    conn.close()

    return rows
```

```python
conditions = {'candidate': ('=', 1),}
select_columns = ['session_id', 'unit_id', 'stimulus']
db_path = r'C:\Users\tmerri03\Desktop\Temp Neural Files\awake_recordings.db'
neurons = load_neurons_from_database(db_path, select_columns, conditions, table='omission_analysis')
```

```python
from polar_analysis import NeuronOmissionAnalysis
import os
import matplotlib.pyplot as plt

recordings_path = r'N:\\Data\\RhythmPerception\\Neural Recordings'

for session_id, unit_id, stimulus in neurons:
    bird = session_id.split(' ')[0]
    recording = os.path.join(recordings_path, bird, session_id)

    analysis = NeuronOmissionAnalysis(
        db_path=db_path,
        stim_lib_path=r'R:\\Data\\tyler\\Recordings\\Stim\\Stimuli Library',
        recordings_path=r'N:\\Data\\RhythmPerception\\Neural Recordings',
        raw_sounds_path=r'C:\\Users\\tmerri03\\Desktop\\RhythmStimuli\\Awake Recs\\Raw Files\\regex',
        recording=recording,
        unit=unit_id,
        stim_subset=[stimulus],
     )
    analysis.run()

    # ---- plotting ----
    fig = plt.figure(figsize=(14, 6))

    # Main layout: left = 2/3, right = 1/3
    outer = fig.add_gridspec(
        1, 2,
        width_ratios=[2, 1],
        wspace=0.3
    )

    # Left side: raster + vector strength stacked
    left = outer[0].subgridspec(
        2, 1,
        height_ratios=[1, 1],
        hspace=0.15
    )

    ax_raster = fig.add_subplot(left[0])
    ax_vs = fig.add_subplot(left[1], sharex=ax_raster)

    # Right side: polar
    ax_polar = fig.add_subplot(
        outer[1],
        projection='polar'
    )

    # Plot
    analysis.analyses[stimulus].plot(
        ax=[ax_raster, ax_vs],
        show=False
    )

    analysis.analyses[stimulus].plot_polar(
        windows=('beat', 'omission', 'decay'),
        ax=ax_polar,
        show=False
    )
    try:
        plt.savefig(rf'C:\Users\tmerri03\Desktop\Neural Data\Awake Plots\Omission Responsive Neurons\{session_id}_{unit_id}_{stimulus}.png', dpi=300)
    except:
        print(f'Failed to save plot for {session_id} {unit_id} {stimulus}')
```
