```python
from pathlib import Path
import hashlib


def file_hash(path, chunk_size=1024 * 1024):
    """Compute SHA256 hash of a file."""
    h = hashlib.sha256()
    with open(path, "rb") as f:
        while chunk := f.read(chunk_size):
            h.update(chunk)
    return h.hexdigest()


def compare_directories(dir1, dir2):
    dir1 = Path(dir1)
    dir2 = Path(dir2)

    # Get all relative file paths
    files1 = sorted(p.relative_to(dir1) for p in dir1.rglob("*") if p.is_file())
    files2 = sorted(p.relative_to(dir2) for p in dir2.rglob("*") if p.is_file())

    # Compare file lists
    only_in_1 = sorted(set(files1) - set(files2))
    only_in_2 = sorted(set(files2) - set(files1))

    if only_in_1:
        print(f"Files only in {dir1}:")
        for f in only_in_1:
            print(f"  {f}")

    if only_in_2:
        print(f"Files only in {dir2}:")
        for f in only_in_2:
            print(f"  {f}")

    # Compare contents of common files
    different = []

    for rel_path in sorted(set(files1) & set(files2)):
        f1 = dir1 / rel_path
        f2 = dir2 / rel_path

        if f1.stat().st_size != f2.stat().st_size:
            different.append(rel_path)
            continue

        if file_hash(f1) != file_hash(f2):
            different.append(rel_path)

    if different:
        print("\nFiles with different contents:")
        for f in different:
            print(f"  {f}")

    if not only_in_1 and not only_in_2 and not different:
        print("Directories are identical.")
        return True
    else:
        print("\nDirectories are NOT identical.")
        return False


```
